```python

import gradio as gr
import random
import re

# Chatbot responses
responses = {
    "greeting": [
        "Hello! How can I help you today?",
        "Hi there! What would you like to know?",
        "Hey! Nice to meet you!"
    ],
    "name": [
        "I am a Smart Chatbot developed using Python.",
        "You can call me Smart Bot!"
    ],
    "python": [
        "Python is a beginner-friendly programming language used in AI, web development and data science."
    ],
    "nlp": [
        "Natural Language Processing (NLP) helps computers understand and process human language."
    ],
    "ai": [
        "Artificial Intelligence enables machines to perform tasks that normally require human intelligence."
    ],
    "thanks": [
        "You're welcome!",
        "Happy to help!"
    ],
    "bye": [
        "Goodbye! Have a great day!",
        "See you again!"
    ]
}

# Chatbot function
def chatbot_response(message, history):

    text = message.lower().strip()
    text = re.sub(r"[^\w\s]", "", text)

    if any(word in text.split() for word in ["hello", "hi", "hey"]):
        return random.choice(responses["greeting"])

    elif "your name" in text or "who are you" in text:
        return random.choice(responses["name"])

    elif "python" in text:
        return random.choice(responses["python"])

    elif "nlp" in text or "natural language" in text:
        return random.choice(responses["nlp"])

    elif "artificial intelligence" in text or " ai " in f" {text} ":
        return random.choice(responses["ai"])

    elif "thank" in text:
        return random.choice(responses["thanks"])

    elif "bye" in text or "goodbye" in text:
        return random.choice(responses["bye"])

    else:
        return (
            "Sorry, I don't understand that question yet. "
            "Try asking me about Python, NLP, AI or technology."
        )

# Create chatbot interface
demo = gr.ChatInterface(
    fn=chatbot_response,
    title="🤖 Smart Text Chatbot",
    description="A simple chatbot developed using Python and Natural Language Processing concepts.",
    examples=[
        "Hello",
        "What is Python?",
        "What is NLP?",
        "What is Artificial Intelligence?",
        "Thank you"
    ]
)

# Launch chatbot
demo.launch(share=True)
```
